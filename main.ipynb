```python
import pandas as pd 
import matplotlib.pyplot as plt

team_data = pd.read_csv("data/team_data.csv")

team_data.info()

# purely scalar/numerical data
columns = [
    "comps_attended",
    "sigs_attended",
    "trueskill",
    "trueskill_ranking",
    "ccwm",
    "opr",
    "dpr",
    "ap_per_match",
    "awp_per_match",
    "wp_per_match",
    "drive_score",
    "auto_score"
]
```

```python
team_data.head()
```

```python
from scipy import stats
from sklearn import linear_model 
import seaborn as sns
import math 

fig, ax = plt.subplots(12, 12)
fig.set_figheight(70)
fig.set_figwidth(70)

for i, x in enumerate(columns):
    for j, y in enumerate(columns):
        if x == y:
            ax[i, j].set_visible(False)
            continue

        x_axis = team_data[x]
        y_axis = team_data[y]


        slope, intercept, r, p, std_err = stats.linregress(x_axis, y_axis)
        
        # model = list(map(lambda x_val: slope * x_val + intercept, x_axis))
        
        # ax.scatter(x_axis, y_axis)
        # ax.plot(x_axis, model)
        # ax.set_title(f"{x} vs. {y}")
        plot_ax = sns.regplot(x=x, y=y, data=team_data, ax=ax[i, j])


        title_text = f"R: {round(r, 4)}, P: {round(p, 4)}, STE: {round(std_err, 3)}"

        if r >= 0.60:
            plot_ax.set_title(title_text, fontweight='bold', color='#66ff00') 
        elif r <= -0.60:
            plot_ax.set_title(title_text, fontweight='bold', color='#FF3131') 
        else:
            plot_ax.set_title(title_text)

plt.savefig("robotics_lin_reg.png",bbox_inches='tight')
plt.show()
```
