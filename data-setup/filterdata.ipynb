```python
import pandas as pd

team_info = pd.read_csv('sq_team_info.csv')

check_nan = [
    'trueskill',
    'trueskill_ranking',
    'opr',
    'dpr',
    'ccwm',
    'ap_per_match',
    'awp_per_match',
    'wp_per_match',
    'drive_score',
    'auto_score'
]
team_info.head()
team_info.info()

```

```python
filtered_team = team_info[team_info['comps_attended'] > 1]
filtered_team.info()
```

```python

filtered_team = filtered_team.dropna(subset=check_nan)
filtered_team.to_csv("team_data.csv", index= False)
filtered_team.info()

```
